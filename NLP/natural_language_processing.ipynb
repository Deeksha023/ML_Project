```python
from nltk.corpus import gutenberg

for fileid in gutenberg.fileids():
    num_chars = len(gutenberg.raw(fileid))
    num_words = len(gutenberg.words(fileid))
    num_sents = len(gutenberg.sents(fileid))
    num_vocab = len(set(gutenberg.words(fileid)))

    print(
        fileid,
        int(num_chars / num_words),
        int(num_words / num_sents),
        int(num_words / num_vocab)
    )
```

```python
text = """Artificial Intelligence is changing the way people work and communicate with technology.
Natural Language Processing helps computers understand, analyze, and generate human language.
Machine learning algorithms can learn useful patterns from large amounts of data and use those patterns to make predictions.
These technologies are widely used in healthcare, education, banking, transportation, and many other fields."""

print("=" * 60)
print("Original text")
print("=" * 60)
print(text)

print("\n" + "=" * 60)
print("Word Tokenization")
print("=" * 60)
words = word_tokenize(text)
print(words)

print("\n" + "=" * 60)
print("Sentence Tokenization")
print("=" * 60)
sentences = sent_tokenize(text)
print(sentences)

print("\n" + "=" * 60)
print("Stop Words Removed")
print("=" * 60)

stop_words = set(stopwords.words("english"))
filtered_words = [word for word in words if word.lower() not in stop_words]

print(filtered_words)
```

```python
# Sentence tokenization
sentences = sent_tokenize(text)

print("\n" + "=" * 60)
print("1. Sentence Tokenization")
print("=" * 60)

for i, s in enumerate(sentences, 1):
    print(f"Sentence {i}: {s}")
```

```python
# Word tokenization
words = word_tokenize(text)

print("\n" + "=" * 60)
print("2. Word Tokenization")
print("=" * 60)
print("Tokens:", words)
print("Number of Tokens:", len(words))
```

```python
# Cleaning (lowercase, remove punctuation and numbers)
cleaned_words = [w.lower() for w in words if w.isalpha()]

print("\n" + "=" * 60)
print("Cleaned Words")
print("=" * 60)
print(cleaned_words)
```

```python
# Remove the stop words
stop_words = set(stopwords.words('english'))

filtered_words = [w for w in cleaned_words if w not in stop_words]

print("\n" + "=" * 60)
print("Filtered Words")
print("=" * 60)
print(filtered_words)
```

```python
stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(w) for w in filtered_words]

print("\n" + "=" * 60)
print("Stemming (Porter Stemmer)")
print("=" * 60)

for orig, stem in zip(filtered_words, stemmed_words):
    print(f"{orig} -> {stem}")
```

```python
import nltk

nltk.download('wordnet')
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(w) for w in filtered_words]

print("\n" + "=" * 60)
print("Lemmatization of Words")
print("=" * 60)

for orig, lemma in zip(filtered_words, lemmatized_words):
    print(f"{orig:15} -> {lemma}")
```

```python
def lemmatize_with_pos(word, tag):
    """Map POS tag to WordNet format and lemmatize."""
    
    tag_map = {
        'J': 'a',
        'N': 'n',
        'V': 'v',
        'R': 'r'
    }
    
    pos = tag_map.get(tag[0], 'n')
    
    return lemmatizer.lemmatize(word, pos=pos)
```

```python
import nltk

nltk.download('averaged_perceptron_tagger_eng')
```

```python
pos_tagged = pos_tag(filtered_words)

lemmatized_pos = [
    lemmatize_with_pos(w, t)
    for w, t in pos_tagged
]

print(lemmatized_pos)
```

JJ – Adjective – Describes a noun

NNP – Proper Noun, Singular – Name of a person, place, organization, etc.

VBZ – Verb, 3rd Person Singular Present – Verb used with he, she, or it

VBG – Verb, Gerund/Present Participle – -ing form of a verb

DT – Determiner – Words like the, a, an

NN – Noun, Singular – A single noun

NNS – Noun, Plural – More than one noun

VBP – Verb, Non-3rd Person Singular Present – Present-tense verb

CC – Coordinating Conjunction – Words like and, but, or

IN – Preposition/Subordinating Conjunction – Words like in, with, on, for

. – Punctuation – Period/full stop
