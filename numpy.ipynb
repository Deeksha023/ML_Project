```python
import sys
cube=[1**3 for i in range(100)]
print("memory allocated for the list is",sys.getsizeof(cube)*len(cube))
```

## np.linspace: this function is similar to arange() function. in this function instead of step size the number of evenly spaced values between the interval is specified
1)start: the starting values of the sequence

2)stop the end value of the sequence

3)num : the number of evenly spaced sample to be generated default is 50

4)restep = if true samples and step between the consecutive number

```python
new_g=np.linspace(1,5,retstep=True)
print(new_g)
print(len(new_g))
```
