```python
import numpy as np
import pandas as pd

np.random.seed(0)

df1 = pd.DataFrame(
    dict(
        time=np.arange(500),
        value=np.random.randn(500).cumsum()
    )
)

df1.head()
```

```python
sns.relplot(x='time', y='value', data=df1, kind='line')
```

```python
df2=pd.DataFrame(np.random.randn(500,2).cumsum(axis=0),columns=['x','y'])
df2.head()
```

```python
sns.relplot(x='x',y='y',data=df2,kind='line',sort=True)
```

```python
sns.relplot(x='x',y='y',data=df2,kind='line',sort=False)
```
