## uses for clustering in wide variety of application:

1)for customer segmentation:- you can cluster customers based on their purchases and their activity on your website. it help to understand the need of the customer 

2)for data analysis clustering help to read the random form of data it make the same type of data as cluster form to read it separetely.

3)an outlier detection:- Any instance that has a low affinity to all the clusters is likely to be an anomly,anomly detection is particularly useful in detecting defects in manufacturing or for fraud detection.

## k-means algorithm:-
kmeans clustering is an unsupervised machine learning algorithm which group the unlabeled dataset into different clusters.here k defines the number of pre-defined clusters thet need to be created in the process as if k=2 then it means there 2 cluster and if it 3, it means there is 3 cluster

## the k means clustering algorithm mainly perform two tasks:-

1)dtermines the best value for k center points or centroids by an iterative process

2)assigns each data points to its closest k center. those data point which are near to the particular k center, that point merge to that cluster

## Elbow method

The Elbow methods is one of the most popular ways to find the optimal number of clusters, this method uses the concept of WCSS value, where WCSS is stands for "within cluster of squares" which defines the total variation within a cluster

```python
plt.scatter(c1['X'],c1['Y'],marker='.')
plt.xlabel("$x_axis$",fontsize=16)
plt.ylabel("$y_axis$",rotation=0,fontsize=16)
plt.title("data in the scatter")
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans=KMeans(n_clusters=2,random_state=42)
```

```python
kmeans.fit(c1)
```

```python
cluster_assignment=kmeans.predict(c1)
```

```python
plt.scatter(c1['X'],c1['Y'],marker='.',c=cluster_assignment)
plt.xlabel("$x_axis$",fontsize=16)
plt.ylabel("$y_axis$",rotation=0,fontsize=16)
plt.title("data in the scatter")
plt.show()
```

```python
c1.head()
```

```python
from sklearn.cluster import KMeans

kmeans_3 = KMeans(n_clusters=3, random_state=42)
```

```python
kmeans_3.fit(c1[['X', 'Y']])
```

```python
kmeans_3.fit(c1)
```

```python
print(kmeans_3.labels_)
```

```python
y_predict=kmeans_3.predict(c1[['X','Y']])
y_predict
```

```python
c1['cluster_prediction']=y_predict
c1.head()
```

```python
c1['cluster_prediction'].value_counts()
```

```python
c1.columns
```

```python
c1['cluster_prediction'].value_counts().plot(
    kind='pie',
    autopct='%.2f%%',
    cmap='Set3'
)

plt.ylabel('')
plt.show()
```

```python
cluster_assignment = cluster_assignment.astype(str)

cluster_assignment[cluster_assignment == '0'] = 'b'
cluster_assignment[cluster_assignment == '1'] = 'r'
cluster_assignment[cluster_assignment == '2'] = 'g'

plt.scatter(c1['X'], c1['Y'], marker='.', c=cluster_assignment)
plt.show()
```

```python
df1 = c1[c1.cluster_prediction == 0]
df2 = c1[c1.cluster_prediction == 1]
df3 = c1[c1.cluster_prediction == 2]
```

```python
plt.scatter(df1.X, df1.Y, color="Pink", marker='.', label='cluster 0')
plt.scatter(df2.X, df2.Y, color="Red", marker='.', label='cluster 1')
plt.scatter(df3.X, df3.Y, color="SkyBlue", marker='.', label='cluster 2')

plt.scatter(
    kmeans_3.cluster_centers_[:, 0],
    kmeans_3.cluster_centers_[:, 1],
    color="Black",
    marker='*',
    s=200,
    label='centroids'
)

plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.show()
```

```python
#plotting of the elbow
sse=[]
k_ran=range(1,10)
for k in k_ran:
    km=KMeans(n_clusters=k)
    km.fit(c1[['X','Y']])
    sse.append(km.inertia_)
plt.xlabel('k')
plt.ylabel("sum of squared error")
plt.plot(k_ran,sse)    
```

```python
c2=pd.read_csv('clustering_data_2.csv')
c2.head(2)
```

```python
plt.scatter(c2['X'],c2['Y'],marker='.')
plt.show()
```

```python
from sklearn.datasets import make_moons
from sklearn.cluster import SpectralClustering,KMeans
from sklearn.metrics import adjusted_rand_score
```

```python
X,y=make_moons(n_samples=300,noise=0.05,random_state=42)
```

```python
sc=SpectralClustering(n_clusters=2,affinity='nearest_neighbors',random_state=42,n_init=10)
```

```python
sc.fit(X)
```

```python
labels_sc=sc.fit_predict(X)
```

```python
km=KMeans(n_clusters=2,random_state=42,n_init=10)
```

```python
labels_km=km.fit_predict(X)
```

```python
#compare and check which fails
print("Spectral ARI:",adjusted_rand_score(y,labels_sc))
print("KMeans ARI :",adjusted_rand_score(y,labels_km))
```

```python
#plotting
fig,axes=plt.subplots(nrows=1,ncols=3,figsize=(15,4))
axes[0].scatter(X[:,0],X[:,1],c=y,cmap='viridis')
axes[0].set_title("ground truth")
axes[1].scatter(X[:,0],X[:,1],c=labels_sc,cmap='viridis')
axes[1].set_title("special clustering")
axes[2].scatter(X[:,0],X[:,1],c=labels_km,cmap='viridis')
axes[2].set_title("k means(fails")
plt.show()
```
