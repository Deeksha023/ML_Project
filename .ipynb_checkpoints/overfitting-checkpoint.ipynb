```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
```

```python
np.random.seed(42)

X = np.linspace(0, 10, 20).reshape(-1, 1)

y = 2 * X.squeeze() + 3 + np.random.normal(0, 2, 20)

print(X)
print(y)
print(X.shape)
```

```python
model = LinearRegression()
model.fit(X, y)
```

```python
model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)

plt.figure(figsize=(8,6))

plt.scatter(X, y, color='red', label="Training data")

plt.plot(X, y_pred, color='blue', linewidth=3, label="Linear Regression")

plt.title("Simple Linear Regression Plotting")
plt.xlabel("X")
plt.ylabel("y")
plt.legend()
plt.grid()

plt.show()
```

```python
poly = PolynomialFeatures(degree=5)

X_poly = poly.fit_transform(X)
```

```python
print("Original X shape:", X.shape)
print("Polynomial X shape:", X_poly.shape)
```

```python
print("\nFirst 5 rows of original X:")
print(X[:5].flatten())

print("\nPolynomial first 5 rows:")
print(np.round(X_poly[:5], 4))
```

```python
degrees=[2,5,10,15]
X_test=np.linspace(0,10,300).reshape(-1,1)
fig,axes=plt.subplots(nrows=2,ncols=2,figsize=(14,14))
axes=axes.ravel()

for i, degree in enumerate(degrees):
    poly=PolynomialFeatures(degree=degree)
    model=LinearRegression()
    X_poly=poly.fit_transform(X)
    model.fit(X_poly,y)
    X_test_poly=poly.transform(X_test)
    y_pred=model.predict(X_test_poly)
    axes[i].scatter(X,y,color="red",label="training data")
    axes[i].plot(X_test,y_pred,color='blue',linewidth=3,label=f'degree{degree}')
    axes[i].set_title(f'plonomail Degree={degree}',fontsize=14)
    axes[i].legend()

plt.show()
```
