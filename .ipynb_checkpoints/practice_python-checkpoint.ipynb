## write a program that prompts the user for two numbers and display the addition,substraction,multiplication, and division betwen them

```python
#prompt the user for two number
num1=float(input("Enter the first number"))
num2=float(input("Enter the second number"))

#perfrom the arithmatic operation
addition=num1+num2
substraction=num1-num2
multiplication=num1*num2
division=num1/num2

#display operation
print("Addition is addtion:",addition)
print("Substraction is:",substraction)
print("Multiplication is:",multiplication)
print("Division is:",division)
```

### write a program that calculates the geometric mean of three numbers entered by the uisers

```python
import math
```

```python
num1=float(input("Enter the first number"))
num2=float(input("Enter the second number"))
num3=float(input("Enter the third number"))
product=num1*num2*num3
g_mean=math.pow(product,1/3)
print("The value of geometric mean is :",g_mean)
```

### write a program  reads three numbers and informs if their sum is divisible by 5 or not

```python
number1 = int(input("Enter the first number: "))
number2 = int(input("Enter the second number: "))
number3 = int(input("Enter the third number: "))

sum_of_number = number1 + number2 + number3

if sum_of_number % 5 == 0:
    print("The sum is divisible")
else:
    print("The sum is not divisible")

```

### writa a program all even numbers from 1-100

```python
num = 1

while num <= 100:
    if num % 2 == 0:
        print(num)
    num = num + 1
```

### write a program that display even numbers 1-50 and odd numbers 51-100 using repeating loop

```python
num=1
while num<=100:
    if num<=50:
        if num%2==0:
            print("Even:",num)
    else:
        if num%2!=0:
            print("Odd:",num)
    num+=1
```

### write a program that asks for two numbers from user and display if the first is divisible by second

### write a program to diplay and find the sum of list numbers using for loop 

```python
l=[10,20,30,40,50]
s=0
for i in l:
    print(i)
    s=s+i
print("Sum of lists is",s)

```

### <font color="red"> Write a program to dipaly the stars in the eqilateral triangle

```python
n = 40

for i in range(1, 11):
    print(' ' * n, end='')
    print('* ' * i)
    n = n - 1
```

### even or odd function

```python
def even_odd(n):
    if n % 2 == 0:
        return "Even"
    else:
        return "Odd"
```

```python
l1=[2,4,3,5,12,13,14,17,16,22,27]
```

```python
print(list(map(even_odd,l1)))
```
