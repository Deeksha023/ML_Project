```python
import ollama

# Reusable function
def ask_llama(question):
    response = ollama.chat(
        model="llama3.2",
        messages=[
            {
                "role": "user",
                "content": question
            }
        ]
    )
    return response["message"]["content"]
```

```python
answer = ask_llama("Explain neural network")
print(answer)
```

```python
def ai_teacher(question):
    response = ollama.chat(
        model="llama3.2",
        messages=[
            {
                "role": "system",
                "content": """You are an expert college lecturer.
Teach students in a simple and practical way.
Explain concepts step by step.
Use real-world examples.
Use Python examples whenever appropriate.
Do not assume advanced knowledge."""
            },
            {
                "role": "user",
                "content": question
            }
        ]
    )

    return response["message"]["content"]
```

```python
question = "What is machine learning?"
answer = ai_teacher(question)
print("Question:", question)
print("\nAI Teacher:")
print(answer)
```

```python
def python_expert(question):
    response = ollama.chat(
        model="llama3.2",
        messages=[
            {
                "role": "system",
                "content": """You are an expert Python programming instructor.

Teach Python to students in a simple and practical way.
Explain concepts step by step.
Use simple language and clear examples.
Provide Python code whenever appropriate.
Explain the code line by line when needed.
Help students find and fix errors in their Python programs.
Explain errors in simple words.
Use real-world examples and practical applications.
Do not assume advanced Python knowledge.
Follow Python best practices.
"""
            },
            {
                "role": "user",
                "content": question
            }
        ]
    )

    return response["message"]["content"]
```

```python
answer = python_expert("Explain Python functions with an example")
print(answer)
```

```python
conversation = []

def chat_with_memory(question):
    
    # Store user's question
    conversation.append({
        "role": "user",
        "content": question
    })
    
    # Send entire conversation to Ollama
    response = ollama.chat(
        model="llama3.2",
        messages=conversation
    )
    
    # Get AI answer
    answer = response["message"]["content"]
    
    # Store AI answer in conversation
    conversation.append({
        "role": "assistant",
        "content": answer
    })
    
    return answer
```

```python
print(chat_with_memory("MY name is Deeksha"))
```

```python
print(chat_with_memory("i am teaching a studnet about generative AI"))
```

```python
response = ollama.chat(
    model="llama3.2",
    messages=[
        {
            "role": "user",
            "content": "Explain Generative AI"
        }
    ],
    stream=True
)

for chunk in response:
    print(chunk["message"]["content"], end="", flush=True)
```

```python
def ask_llama_stream(question):
    response = ollama.chat(
        model="llama3.2",
        messages=[
            {
                "role": "user",
                "content": question
            }
        ],
        stream=True
    )

    for chunk in response:
        print(chunk["message"]["content"], end="", flush=True)

    print()
```

```python
ask_llama_stream("Explain the karnatka politics")
```
