```python
import ollama
```

```python
conversation = [
    {
        'role': 'system',
        'content': """
        you are a generative AI teacher
        explain concepts clearly and step by step
        """
    }
]

def astro_talk(question):
    conversation.append({
        'role': 'user',
        'content': question
    })

    response = ollama.chat(
        model='llama3.2',
        messages=conversation,
        stream=True
    )

    complete_answer = ""

    for chunk in response:
        text = chunk['message']['content']
        print(text, end='', flush=True)
        complete_answer = complete_answer + text

    conversation.append({
        'role': 'Guruji',
        'content': complete_answer
    })
```

```python
astro_talk("Which is best palm astrology or name astrology")
```
